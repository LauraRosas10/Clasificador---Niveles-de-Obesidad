## **Contextualización**

**Modelo**

Se llama modelo porque es una representación matemática o computacional que relaciona un conjunto de variables de entrada (como peso, altura, edad, hábitos alimentisios, actividad física) con una variable de salida (el nivel de obesidad).

En otras palabras:

•	Recibe datos → peso, altura, actividad física, etc.

•	Aprende patrones a partir de ejemplos previos (personas con su nivel de obesidad conocido).

•	Predice resultados → clasifica nuevas personas en categorías de obesidad.

Es un modelo porque abstrae la realidad, captura relaciones importantes entre variables y permite hacer predicciones basadas en datos, sin necesidad de medir o analizar manualmente cada caso.


**Problemática**

La obesidad es un problema de salud pública creciente a nivel mundial, asociada a enfermedades crónicas como diabetes, hipertensión, problemas cardiovasculares y ciertos tipos de cáncer. Muchas personas no son conscientes de su nivel de riesgo hasta que desarrollan complicaciones graves.

Además, los métodos tradicionales de diagnóstico, basados solo en peso o índice de masa corporal (IMC), no siempre reflejan de manera precisa el riesgo individual, ya que no consideran factores como hábitos alimenticios, actividad física, genética o estilo de vida.

Esta falta de información precisa dificulta la prevención temprana y la toma de decisiones personalizadas en salud, limitando la capacidad de diseñar planes de intervención efectivos.

Por ello, aplicar un modelo de predicción del nivel de obesidad permite clasificar a los individuos según su riesgo real, integrando múltiples variables, y así apoyar estrategias de prevención, control y cuidado de la salud de manera más eficiente y personalizada.



 **Niveles de Obesidad**

 La clasificacion de los niveles de obesidad en la actualidad se obtienen de la siguiente forma:




image.png

La forma más común de determinar si una persona tiene sobrepeso u obesidad es calcular su índice de masa corporal (BMI por sus siglas en inglés) El BMI es una estimación de la grasa corporal que se obtiene al comparar el peso de una persona con su estatura.

Los profesionales de la salud también observan el BMI, junto con información sobre factores de riesgo adicionales, para determinar el riesgo de una persona de desarrollar enfermedades relacionadas con el peso. En general, cuánto más alto es el BMI de una persona, mayor es el riesgo de tener enfermedades.

<br/>

***Índice de masa corporal para adultos.***

 Los profesionales de la salud utilizan los rangos de BMI para indicar la situación de una persona en cuanto al peso. Para los adultos, un BMI de:

- 18,5 a 24,9 se considera peso normal.
- 25,0 a 29,9 se considera sobrepeso.
- 30,0 a 34,9 se considera obesidad tipo I.
- 35 a 39,9 se considera obesidad tipo II.
- 40,0 o más se considera obesidad tipo III (extrema).

Es importante recordar que, si bien el BMI suele ser adecuado para estimar cuánta grasa corporal tiene una persona, no mide la grasa corporal directamente y, por lo tanto, no es confiable en todos los casos. *Por ejemplo, una persona podría pesar más porque es atlética y tiene mucha masa muscular, y no porque tenga demasiada grasa corporal.*

<br/>


Además, la grasa corporal se pude medir de otras maneras, como la circunferencia de la cintura, el índice cintura-cadera, el grosor de los pliegues cutáneos (un pellizco de piel y grasa) y técnicas como el ultrasonido, que son más precisas que el BMI. Un profesional de la salud puede ayudar a determinar si es necesario realizar estas pruebas.


Adicionalmente, la obesidad puede clasificarse según la distribución de la grasa corporal:

- **Obesidad androide (forma de manzana)**: acumulación de grasa en abdomen y torso.  
- **Obesidad ginoide (forma de pera)**: acumulación de grasa en caderas y muslos.  

Esta clasificación es fundamental para evaluar riesgos de salud y diseñar intervenciones personalizadas.

<br/>

**Usuarios-Beneficio**

***El Público General (Ciudadanos Comunes)***:

El mayor beneficio es: el modelo proporciona una evaluación de riesgo individual antes de que la obesidad se desarrolle o se agrave.

- Un individuo puede ingresar sus datos de estilo de vida (hábitos alimenticios, frecuencia de ejercicio, uso de tecnología, etc.) y obtener una predicción de su nivel de riesgo de obesidad (Sobrepeso, Saludable, Obesidad tipo I, etc ), siendo esto un llamado de atención, informando al usuario que sus hábitos actuales lo están llevando a un nivel de riesgo específico.

<br/>

***Validación de la Necesidad del IMC (Variable Combinada)***


**Superposición de Altura (Eje X):** Los rangos de altura para todas las clases de obesidad se superponen ampliamente (la mayoría de las personas caen entre 1.6 m y 1.8 m, independientemente de su nivel de obesidad). Esto significa que la altura por sí sola no es un buen predictor.

***El Factor Clave es la Relación (IMC):*** La superposición de alturas, combinada con la separación vertical por peso, demuestra por qué es esencial la métrica IMC = peso (kg) / altura² (m²).

Los niveles de obesidad no dependen del peso o la altura individualmente, sino de cuánto peso tiene una persona para su altura.

Una persona en la esquina inferior derecha (alta y con bajo peso) estará en la categoría de "Peso Normal" o "Insuficiente".
El IMC ya usa la altura para normalizar el peso.

#VARIABLES TOTALES-NUMERICAS "Mapa de Calor"

```python
#   columnas categóricas
#df=df.drop("SMOKE",axis=1)
from sklearn.preprocessing import LabelEncoder
categorical_cols = df.select_dtypes(include=['object']).columns
print("Columnas categóricas:", categorical_cols.tolist())

# Diccionario para guardar los mapeos
mapeos = {}

# Transformar cada columna categórica y guardar el mapeo
for col in categorical_cols:
    le = LabelEncoder()
    df[col] = df[col].fillna('missing')  # para manejar nulos
    df[col] = le.fit_transform(df[col])
    mapeos[col] = dict(zip(le.classes_, le.transform(le.classes_)))

# Mostrar los mapeos
print("\nMapeos de columnas categóricas:\n")
for col, mapping in mapeos.items():
    print(f"Columna: {col}")
    for categoria, valor in mapping.items():
        print(f"  {categoria}: {valor}")
    print("-" * 40)

    # Selecciona solo columnas numéricas
numeric_cols = df.select_dtypes(include=['int64', 'float64'])

#Matriz de correlación
corr_matriz = numeric_cols.corr()

# Mostrar la matriz de correlación
plt.figure(figsize=(12, 8))
sns.heatmap(corr_matriz, annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Mapa de correlación de variables")
plt.show()
```
